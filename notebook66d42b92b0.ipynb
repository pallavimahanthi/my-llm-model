```python
!pip -q install transformers torch

from transformers import GPT2Config, GPT2LMHeadModel

# Create a tiny language model from scratch
config = GPT2Config(
    vocab_size=1000,
    n_positions=64,
    n_embd=64,
    n_layer=2,
    n_head=2,
    bos_token_id=None,
    eos_token_id=None
)

model = GPT2LMHeadModel(config)

print("My own small LLM model created!")
print("Total parameters:", sum(p.numel() for p in model.parameters()))
```

```python
import torch

model.eval()

input_ids = torch.tensor([[1, 10, 20]])

with torch.no_grad():
    output = model(input_ids)

print("Output shape:", output.logits.shape)
print("My LLM model is working!")
```

```python
# Save your LLM model
model.save_pretrained("/kaggle/working/my_llm_model")

print("Model saved successfully!")
```
